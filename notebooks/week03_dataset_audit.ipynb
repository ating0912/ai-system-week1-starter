## 6. Split and Leakage Notes

- This dataset has no `timestamp`, so it cannot support a time split.
- This dataset has no `user_id` or `ticket_id`, so it cannot support a group split.
- `intent` is the target and must not be used as input.
- If future versions add `resolution`, `final_case_code`, or `refund_completed_at`, those fields may create future or target-derived leakage.

## 7. Audit Conclusion

The dataset is clean enough for a Week 3 teaching demo, but it is synthetic, balanced, and lacks time / user / ticket metadata. It should not be treated as evidence for real deployment performance.
